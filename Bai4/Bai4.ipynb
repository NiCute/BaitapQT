Phân tích đơn biến (Univariate Analysis)

```python
plt.figure(figsize=(6, 4))
sns.histplot(data=df, x='G3', kde=True)
plt.title('phan bo diem G3 nam va nu')
plt.show()

mean = round(df['G1'].mean(), 2)
median = round(df['G1'].median(), 2)
mode = df['G1'].mode()[0]

print("Mean G1  :", mean)
print("Median G1:", median)
print("Mode G1  :", mode)

# So sánh theo 3 mẫu
if abs(mean - median) <= 1 and abs(median - mode) <= 1:
    print("=> Đồ thị thuộc loại: Chuẩn, đối xứng (Normal)")
elif mode < median < mean:
    print("=> Đồ thị thuộc loại: Lệch phải (Right-skewed)")
elif mean < median < mode:
    print("=> Đồ thị thuộc loại: Lệch trái (Left-skewed)")

```
